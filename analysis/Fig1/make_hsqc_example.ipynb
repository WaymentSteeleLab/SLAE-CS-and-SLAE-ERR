```python
import os
os.makedirs("Figs", exist_ok=True)  # figures are written to ./Figs

import sys
MAKESHIFT_ROOT = "../../external/makeshift"
if MAKESHIFT_ROOT not in sys.path:
    sys.path.insert(0, MAKESHIFT_ROOT)

from makeshift.spectra import Spectrum
from makeshift.spectra.plotting import plot_spectrum, plot_peaklist

hsqc = Spectrum.example_hsqc()            # or Spectrum.from_ucsf("my.ucsf")
peaks = hsqc.pick_peaks(baseline=20)
fig, ax = plot_spectrum(hsqc, xlim=(9.5, 6.5), ylim=(130, 105),
                        cmap='Greys_r', baseline=20)
# plot_peaklist(peaks, ax=ax, text=None, color='red')
```

```python
import matplotlib as mpl
mpl.rcParams['svg.fonttype'] = 'none'   # keep text as editable text (not paths) in Illustrator/Inkscape
mpl.rcParams.update({
    'font.family': 'sans-serif',
    'font.sans-serif': ['Arial'],
    # make $...$ math text (the 1H / 15N superscripts) use Arial too
    'mathtext.fontset': 'custom',
    'mathtext.rm': 'Arial',
    'mathtext.it': 'Arial:italic',
    'mathtext.bf': 'Arial:bold',
    'svg.fonttype': 'none',   # keep text editable in the exported SVG
})


fig, ax = plot_spectrum(hsqc, xlim=(9, 7.5), ylim=(125, 115),
                         baseline=20,
                        figsize=(7, 8), contour_color='green')   # width, height in inches

ax.xaxis.label.set_fontsize(20)     # 1H (ppm)
ax.yaxis.label.set_fontsize(20)     # 15N (ppm)
ax.tick_params(labelsize=20)        # tick numbers
from matplotlib.ticker import MultipleLocator

ax.xaxis.set_major_locator(MultipleLocator(0.5))
import numpy as np
ax.set_yticks(np.arange(115, 126, 2))   # 115, 117, ..., 125

fig.savefig('Figs/hsqc_example.svg', format='svg', bbox_inches='tight', transparent=True)
```
