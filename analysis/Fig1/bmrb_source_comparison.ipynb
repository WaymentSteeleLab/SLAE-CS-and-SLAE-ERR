```python
import os
os.makedirs("Figs", exist_ok=True)  # figures are written to ./Figs

import matplotlib.pyplot as plt
import numpy as np

def set_style(base=12, family="Arial"):
    plt.rcParams.update({
        "font.family": family,
        "font.size": base,
        "axes.titlesize": base + 2,
        "axes.labelsize": base,
        "xtick.labelsize": base - 2,
        "ytick.labelsize": base - 2,
        "legend.fontsize": base - 2,
        "svg.fonttype": "none",
    })

set_style(base=17)

labels = [
    "Whole BMRB", "RefDB", "rBMRB"
]

means = [15981, 2429, 6123]

# colour by training-data group
colors = (["#D2D9E0"] * 1) + (["#98C379"] * 1) + (["#61AFEF"] * 1)

x = np.arange(len(labels))

fig, ax = plt.subplots(figsize=(4.5, 5))

bars = ax.bar(x, means, color=colors,
            edgecolor="white", linewidth=0.6)

# value labels
for bar, m, lab in zip(bars, means, labels):
    ax.text(bar.get_x() + bar.get_width() / 2, m + max(means) * 0.01,
            f"{m:,.0f}", ha="center", va="bottom",
            fontweight="bold" if lab == "rBMRB" else "normal")

ax.set_xticks(x)
ax.set_xticklabels(labels)
for t in ax.get_xticklabels():
    t.set_fontweight("bold" if t.get_text() == "rBMRB" else "normal")


ax.set_ylabel("Number of protein entries")
ax.set_ylim(0, max(means) * 1.15)
# ax.set_title("Entries per Source Database", fontweight="bold")

# ax.grid(axis="y", alpha=0.3, linestyle="--")
plt.tight_layout()
plt.savefig("Figs/bmrb_source_comparison.png", dpi=300, bbox_inches="tight")
plt.savefig("Figs/bmrb_source_comparison.svg", bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def set_style(base=12, family="Arial"):
    plt.rcParams.update({
        "font.family": family,
        "font.size": base,
        "axes.titlesize": base + 2,
        "axes.labelsize": base,
        "xtick.labelsize": base - 2,
        "ytick.labelsize": base - 2,
        "legend.fontsize": base - 2,
        "svg.fonttype": "none",
    })

set_style(base=20)

labels = [
    "Whole BMRB", "RefDB", "rBMRB\n(this work)"
]

means = [15981, 2429, 6123]

# colors grouped by training data
colors = (["#D2D9E0"] * 1) + (["#A9A9A9"] * 1) + (["#61AFEF"] * 1)

y = np.arange(len(labels))

fig, ax = plt.subplots(figsize=(11, 4))

bars = ax.barh(y, means, color=colors,
            edgecolor="white", linewidth=0.6)

# value labels
for bar, m, lab in zip(bars, means, labels):
    ax.text(bar.get_width() + max(means) * 0.01, bar.get_y() + bar.get_height() / 2,
            f"{m:,.0f}", ha="left", va="center", fontweight="bold" if lab == "rBMRB\n(this work)" else "normal")

ax.set_yticks(y)
ax.set_yticklabels(labels)
for tick_label, lab in zip(ax.get_yticklabels(), labels):
    if lab == "rBMRB\n(this work)":
        tick_label.set_fontweight("bold")
ax.invert_yaxis()  # keep the first label on top
ax.set_xlabel("Number of protein entries")
ax.set_xlim(0, max(means) * 1.15)
ax.set_xticks(np.arange(0, max(means) * 1.2, step=4000))
# ax.set_title("Entries per Source Database", fontweight="bold")

# ax.grid(axis="x", alpha=0.3, linestyle="--")
plt.tight_layout()
plt.savefig("Figs/bmrb_source_comparison_h.png", dpi=300, bbox_inches="tight")
plt.savefig("Figs/bmrb_source_comparison_h.svg", bbox_inches="tight")
plt.show()
```
